```python
import random

class Cliente:
    """ Definimos nuestra clase donde inmediatamente agregamos uno de los
    supuestos """
    def __init__(self, tiempo_llegada, cliente_premier):
        self.tiempo_llegada  = tiempo_llegada
        self.cliente_premier = cliente_premier
        #el tiempo que tarda en atender un operador a un cliente
        self.tiempo_servicio = random.uniform(1, 81)

class Cola1:
    """Definimos nuetras dos colas en variables como listas vacías, los tiempos
       en cero para ir actualizando y una variable con una lista que contiene
       al cero n-veces el valor del num_lineas"""
    def __init__(self, num_lineas):
        self.num_lineas = num_lineas
        self.lineas = [0] * num_lineas
        self.cola = []
        self.cola_premier = []
        self.tiempo_actual = 0
        self.tiempo_espera = 0
        self.total_clientes = 0
        self.max_premier = 0

    def meter_cola(self, cliente):
        """ Agrega los elementos a la cola, dependiendo del tipo de cliente
        los manda a la cola premier o a la cola normal """
        if cliente.cliente_premier:
            self.cola_premier.append(cliente)
        else:
            self.cola.append(cliente)

    def procesar_llamada(self):
        """ itera sobre el rango de num_lineas para ir sacanto los elementos
        de la cola dependiendo, condiciona si el valor con el indice es menor
        que el tiempo actual y además si es cliente premier priorisa su salida,
        luego, va actializando los valores de las variables en cero """
        for i in range(self.num_lineas):
            if self.lineas[i] <= self.tiempo_actual:
                if self.cola_premier:
                    cliente = self.cola_premier.pop(0)
                elif self.cola:
                    cliente = self.cola.pop(0)
                else:
                    continue

                tiempo_espera_cliente = (self.tiempo_actual) \
                                         - (cliente.tiempo_llegada)
                self.tiempo_espera += tiempo_espera_cliente
                self.total_clientes += 1

                if cliente.cliente_premier:
                    self.max_premier = (max(self.max_premier, \
                                               tiempo_espera_cliente))

                self.lineas[i] = self.tiempo_actual + cliente.tiempo_servicio

    def simulacion(self, clientes):
        """ Iteramos sobre cada uno de los clientes propuestos el cual entra en
        un ciclo while el cual depende de que el valor del tiempo de llegada
        sea mayor al del tiempo actual para entonces "llamar" a la función que
        procesa la llamada y se actualiza el tiempo actual y se """
        for cliente in clientes:
            while cliente.tiempo_llegada > self.tiempo_actual:
                self.procesar_llamada()
                self.tiempo_actual += 1

            self.meter_cola(cliente)

        while self.cola_premier or self.cola:
            self.procesar_llamada()
            self.tiempo_actual += 1

        promedio_tiempo_espera = (self.tiempo_espera / self.total_clientes \
         if self.total_clientes > 0 else 0)

        return promedio_tiempo_espera, self.max_premier

def generador_clientes(num_clientes):
    """Generamos una cola de clientes donde el tiempo de llegada de cada
    cliente este en un intervalo de 1 a 3, asi como 1/6 de los clientes sean
    premier, cada cliente tendra el tiempo de llegada y si es premier o no"""
    clientes = []
    tiempo_actual = 0
    for _ in range(num_clientes):
        tiempo_actual += random.uniform(1, 3)
        cliente_premier = random.random() < 1/6
        clientes.append(Cliente(tiempo_actual, cliente_premier))
    return clientes

def encontrar_lineas_necesarias(espera_promedio_ideal, clientes):
    """Para encontara el minimo de lineas necesarias se inicia con una linea
    se ejecuta la simulacion, se obtiene el tiempo promedio de espera y el
    tiempo maximo de espera para los premier. Si el tiempo de espera es menor
    o igual al tiempo de espera objetivo da el número de lineas, el promedio
    de espera y el maximo de espera premier, si no se agrega una linea hasta
    alcanzar el objetivo de tiempo de espera"""
    num_lineas = 1
    while True:
        llamadas = Cola1(num_lineas)
        promedio_espera, max_premier = (llamadas.simulacion\
                                              (clientes))
        if max_premier <= espera_promedio_ideal:
            return num_lineas, max_premier, promedio_espera
        num_lineas += 1

if __name__ == "__main__":
 num_clientes = 10000 #número de clientes propuestos
 espera_promedio_ideal = 10

 clientes = generador_clientes(num_clientes)
 lineas_necesarias, promedio_espera, max_premier = (encontrar_lineas_necesarias\
                                         (espera_promedio_ideal,clientes))

 print(f"Número de clientes: {num_clientes}")
 print(f"Numero de lineas necesarias: {lineas_necesarias}")
 print(f"Promedio de tiempo de espera: {promedio_espera:.2f} min.")
 print(f"Tiempo máximo de espera para clientes premier: {max_premier:.2f} min.")
```
